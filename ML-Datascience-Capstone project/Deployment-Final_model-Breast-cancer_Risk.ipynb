```python
import pickle
```

```python
loaded_model=pickle.load(open('cancerpredict_finalmodel.sav','rb'))

def get_binary_input(prompt):
    val = input(prompt + " (Yes/No): ").strip().lower()
    return 1 if val == 'yes' else 0

print("--- Breast Cancer Risk Assessment Tool ---")

Family_History = get_binary_input("Family history of breast cancer?")
BMI = float(input("Enter BMI: "))
High_Fat_Diet = get_binary_input("High fat diet?")
Smoking = get_binary_input("Current smoker?")
Alcohol = get_binary_input("Frequent alcohol consumption?")
HRT_Usage = get_binary_input("Hormonal Replacement Therapy usage?")
Late_Menopause = get_binary_input("Late menopause?")
No_or_Late_Pregnancy = get_binary_input("Late or no pregnancy?")
Breastfed = get_binary_input("Breastfed children?")
Estrogen_Cosmetics = get_binary_input("Use of estrogen-based cosmetics?")

import pandas as pd
user_data = pd.DataFrame([[Family_History, BMI, High_Fat_Diet, Smoking, Alcohol, 
                           HRT_Usage, Late_Menopause, No_or_Late_Pregnancy, 
                           Breastfed, Estrogen_Cosmetics]], 
                         columns=['Family_History', 'BMI', 'High_Fat_Diet', 'Smoking', 'Alcohol', 
                                  'HRT_Usage', 'Late_Menopause', 'No_or_Late_Pregnancy', 
                                  'Breastfed', 'Estrogen_Cosmetics'])

prediction = loaded_model.predict(user_data.values)[0]
```

```python
risk_map={0:"Low",1:"Moderate",2:"High"}
print(f"Prediction: The breast cancer risk level is {risk_map[prediction]}")
```
