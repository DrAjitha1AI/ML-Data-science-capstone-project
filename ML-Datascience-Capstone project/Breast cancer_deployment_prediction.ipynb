```python
import pickle
```

```python
loaded_model=pickle.load(open('cancerpredict_rfmodel.sav','rb'))


fam_history_text=input("Do you have a family history of breast cancer? (Yes/No):")

if fam_history_text.strip().lower() =='yes':
    fam_hist_numeric =1 
else:
    fam_hist_numeric =0

bmi_input = float(input("Enter BMI: "))

import pandas as pd
user_data=pd.DataFrame([[fam_hist_numeric, bmi_input]], columns=['Family_History','BMI'])

prediction=loaded_model.predict(user_data)
```

```python
risk_map={0:"Low",1:"Moderate",2:"High"}
print(f"Prediction: The breast cancer risk level is {risk_map[prediction[0]]}")
```
